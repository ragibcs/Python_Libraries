```python
# Movies Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
#plot
plt.plot(oscar_years,oscar_revenue)
plt.title("Oscar movies revenue")
plt.xlabel("year")
plt.ylabel("revenue")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
plt.plot(oscar_years,oscar_revenue,"o--r",label="Oskar Winners")
plt.plot(non_oscar_years,non_oscar_revenue,"<-y",label="Non- Oskar Winners")
plt.title("Oskar movies revenue vs Non Oskar revenue")
plt.xlabel("years")
plt.ylabel("Revenue")
plt.legend()
```

```python
with plt.xkcd():    
        plt.plot(oscar_years,oscar_revenue,label="Oskar Winners")
        plt.plot(non_oscar_years,non_oscar_revenue,label="Non- Oskar Winners")
        plt.title("Oskar movies revenue vs Non Oskar revenue")
        plt.xlabel("years")
        plt.ylabel("Revenue")
        plt.legend()
        # print(plt.style.available)
        # plt.style.use("default")
        # plt.grid()
        plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
oskar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oskar_years = [2008, 2009, 2010, 2011, 2012]
oskar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

x=np.arange(len(oskar_years))

width=0.4
plt.bar(x-width/2,oskar_revenue,width,label="Oskar Movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="Non Oskar movies")

plt.title("Oscar vs Non_Oskar movies")
plt.xlabel("Years")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(x,oskar_years )
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
# colors=["red","purple","yellow","blue","pink","green","blue","black","pink","blue"]
colors=["green" if x< 135 else "red" for x in bp] 
sizes=bp
# plt.scatter(age,bp,color=colors,s=bp,marker="o",alpha=0.2)
plt.scatter(age,bp,cmap="OrRd",c=bp)
plt.title("Age vs BP")
plt.xlabel("age")
plt.ylabel("BP")
plt.grid()
plt.colorbar()
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],bp[i]),xytext=(age[i],bp[i]))
plt.xlim(min(age), max(age)+10)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp,winter_humidity,color="blue",label="winter")
plt.scatter(summer_temp,summer_humidity,color="orange",label="Summer")

plt.xlabel("Temperature in (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

# plt.style.use("ggplot")
# plt.pie(amounts,labels=expenses, autopct='%1.1f')

colors = ["purple", "green", "blue", "pink", "yellow"]
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', colors=colors)

# wedgeprops
plt.pie(amounts, labels=expenses,  autopct='%1.1f%%', 
        wedgeprops={"edgecolor":"black", "linewidth": 1, "linestyle":"-.", "fill":False}
       )
```
